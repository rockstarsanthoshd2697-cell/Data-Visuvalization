```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[2,4,6,8,10]
x1=[1,2,3,4]
y1=[2,6,3,9]
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.xlim(1,10)
plt.ylim(5,15)
plt.title("Line plot example")
plt.plot(x,y,label="line-1",color='black',linestyle='dashed',linewidth=3,marker='o',markerfacecolor='grey',markersize=15)
plt.plot(x1,y1,label="line-2")
plt.legend()
plt.savefig("sample dv.png")
plt.show()
```
